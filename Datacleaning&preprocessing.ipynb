```python
df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
df["price_clean"] = (
    df["price"]
    .astype(str)
    .str.replace("Â£", "", regex=False)
    .str.strip()
)
```

```python
print(df[["price", "price_clean"]].head(10))
```

```python
df["price_clean"] = pd.to_numeric(
    df["price_clean"],
    errors="coerce"
)
```

```python
print(df[["price", "price_clean"]].head(10))
```

```python
print(df["price_clean"].isnull().sum())
```

```python
print(df["price_clean"].dtype)
```

```python
print(df["rating"].unique())
```

```python
rating_mapping = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}
```

```python
df["rating_numeric"] = df["rating"].map(rating_mapping)
```

```python
df[["rating", "rating_numeric"]].head(10)
```

```python
print(df["rating_numeric"].isnull().sum())
```

```python
print(df["rating_numeric"].isnull().sum())
```

```python
print(df["availability"].unique())
```

```python
df["availability"] = (
    df["availability"]
    .str.strip()
    .str.lower()
)
```

```python
print("Minimum price:", df["price_clean"].min())
print("Maximum price:", df["price_clean"].max())
```

```python
def price_category(price):
    if price < 20:
        return "Low"
    elif price < 40:
        return "Medium"
    else:
        return "High"
```

```python
df["price_category"] = df["price_clean"].apply(price_category)
```

```python
df[["price_clean", "price_category"]].head(10)
```

```python
df["price_category"].value_counts()
```

```python
def rating_category(rating):
    if rating >= 4:
        return "Highly Rated"
    elif rating >= 3:
        return "Average"
    else:
        return "Low Rated"
```

```python
df["rating_category"] = df["rating_numeric"].apply(rating_category)
```

```python
df[["rating_numeric", "rating_category"]].head(10)
```

```python
df["rating_category"].value_counts()
```

```python
print(df["price_category"].isnull().sum())
print(df["rating_category"].isnull().sum())
```
